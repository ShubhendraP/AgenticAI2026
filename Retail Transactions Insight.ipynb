## Task 3: Customer Behaviour Analysis

```python
# Which customer categories spend the most on average?

#import pandas as pd

# Read your CSV
#df = pd.read_csv("Retail_Transactions_Dataset.csv")

# Convert Total_Cost to numeric (if needed)
df['Total_Cost'] = pd.to_numeric(df['Total_Cost'], errors='coerce')

# Group by Customer_Category, calculate average spend
avg_spend = df.groupby('Customer_Category')['Total_Cost'].mean().sort_values(ascending=False)
print(avg_spend.head(5))
```

```python
# Do certain customer categories prefer specific payment methods?

#import pandas as pd
# Read CSV
#df = pd.read_csv("Retail_Transactions_Dataset.csv")

# Create crosstab: rows=Customer_Category, columns=Payment_Method, values=transaction counts
payment_preferences = pd.crosstab(
    df['Customer_Category'], 
    df['Payment_Method'], 
    normalize='index'  # percentages (0-1) per category
)

print(payment_preferences.round(3))  # Top preferences stand out

# The below output shows the below trends:
#Middle-Aged  prefer Mobioe payment
#Professional  prefer Cash
#Retirees prefer Credit card and Mobile payment
#Senior citizen prefer cash
#Student prefer cash and debit card
#Teenager prefer credit card
#Young adult prefer debit card
```

```python
# What is the average number of items bought per transaction per store type?
import pandas as pd

# Read CSV
#df = pd.read_csv("Retail_Transactions_Dataset.csv")

# Convert to numeric (handles any messy data)
df['Total_Items'] = pd.to_numeric(df['Total_Items'], errors='coerce')

# Group by Store_Type, calculate average items per transaction
avg_items = df.groupby('Store_Type')['Total_Items'].mean().round(2)
print(avg_items.sort_values(ascending=False))
```

## Task 4: Promotion & Discount Impact

```python
#What is the average cost of transactions where a discount was applied vs not applied?

import pandas as pd

#df = pd.read_csv("Retail_Transactions_Dataset.csv")
df['Total_Cost'] = pd.to_numeric(df['Total_Cost'], errors='coerce')

# Find the avg cost of transactions as "No Discount" vs "Discount" and relabel the result columns
avg_cost = (df.groupby('Discount_Applied')['Total_Cost']
            .mean()
            .rename({False: 'No Discount', True: 'Discount Applied'})
            .round(2))
print(avg_cost)
```

## Task 5: Seasonality Trends

```python
#Which season has the highest total revenue?

df['Total_Cost'] = pd.to_numeric(df['Total_Cost'], errors='coerce')

#find the revenue against each season in descending order
total_cost = (df.groupby('Season')['Total_Cost']
                .sum()
                 .round().sort_values(ascending=False))
print("Total Revenue by season:")
print (total_cost)
#print(f"\nHighest: {total_cost.index[0]} (${total_cost.iloc[0]:,.0f})")

#Filter the data to show the season with highest revenue be selecting the t
print(f"\nHighest", total_cost.head(1))

```

```python
# Are there seasonal preferences for certain store types or product categories?

#import pandas as pd
#import ast

# Read and prepare data
#df = pd.read_csv("Retail_Transactions_Dataset-Copy.csv")
df['Total_Cost'] = pd.to_numeric(df['Total_Cost'], errors='coerce')

# 1. Season vs Store_Type preferences (crosstab %)
print("=== Store Type by Season (Top Preferences) ===")
store_season = pd.crosstab(df['Season'], df['Store_Type'], normalize='index').round(3)
display(store_season.sort_values(by=store_season.index[0], axis=1, ascending=False))

print("\n=== Products by Season (Top 3 per season) ===")
# 2. Explode products for seasonal analysis
#df['Product'] = df['Product'].apply(ast.literal_eval) # not running this again as the list is already stored as a python list
products_exploded = df.explode('Product')

# Top products per season
top_products_season = products_exploded.groupby(['Season', 'Product']).size().reset_index(name='count')
top3products_per_season = (top_products_season
                   .sort_values(['Season', 'count'], ascending=[True, False])
                   .groupby('Season')
                   .head(3))
print(top3products_per_season)
```

```python
#Line chart of monthly revenue trends (grouped by year if applicable)
import pandas as pd
import matplotlib.pyplot as plt

# Read data
df['Total_Cost'] = pd.to_numeric(df['Total_Cost'], errors='coerce')


# Extract year-month
#df['month'] = df['Date'].dt.month('M') # already extracted above and converted from datetime to a date value

# Monthly revenue
monthly_revenue = df.groupby('month')['Total_Cost'].sum().round(0)

# Line chart
plt.figure(figsize=(14, 6))
monthly_revenue.plot(kind='line', marker='o', linewidth=2)

plt.title("Monthly Revenue Trends")
plt.xlabel("Year-Month")
plt.ylabel("Total Revenue ($)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
## Heatmap or stacked bar showing revenue by season and customer category

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read data
df['Total_Cost'] = pd.to_numeric(df['Total_Cost'], errors='coerce')

# Revenue crosstab
revenue_heatmap = pd.crosstab(df['Season'], df['Customer_Category'], 
                             values=df['Total_Cost'], aggfunc='sum').round(0)

# Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(revenue_heatmap, annot=True, fmt='.0f', cmap='Blues', cbar_kws={'label': 'Revenue'}) # using fmt = .0f for rounding off
plt.title("Revenue Heatmap: Season vs Customer Category")
plt.xlabel("Customer Category")
plt.ylabel("Season")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read data
df['Total_Cost'] = pd.to_numeric(df['Total_Cost'], errors='coerce')

# Pivot for stacked bar (Season on x-axis, categories stacked)
revenue_stacked = df.pivot_table(
    values='Total_Cost', 
    index='Season', 
    columns='Customer_Category', 
    aggfunc='sum', 
    fill_value=0
).round(0)

# Stacked bar chart
revenue_stacked.plot(kind='bar', stacked=True, figsize=(12, 7), width=0.8)

plt.title("Revenue by Season and Customer Category (Stacked)")
plt.xlabel("Season")
plt.ylabel("Total Revenue ($)")
plt.legend(title="Customer Category", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
