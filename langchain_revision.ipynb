```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

# Create our model

llm = ChatOpenAI(model_name='gpt-4o-mini')

prompt = ChatPromptTemplate.from_template(
    """Wite a python program and write test cases as well for the program.
    User Input : {input} """

)

chain = prompt | llm

#response = chain.invoke({"Write a program to test whether a number is prime or not"})

response = chain.invoke({"Write a program to test whether a number is odd or even"})

print(response.content)




```

```python
from langchain_core.prompts import PromptTemplate
template = PromptTemplate(input_variables= ["topic"],
                         template = """Write me a blog post on {topic}"""
                         )

#prompt = template.format(topic='AI')

chain = template | llm

response = chain.invoke({'topic':'MAchine Leanrning'})

print(response.content)

```
