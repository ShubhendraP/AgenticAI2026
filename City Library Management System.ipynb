```python
# -----------------------------
# City Library Management System
# -----------------------------

# Data structures
books = {}        # book_id -> {"title":..., "author":..., "genre":..., "available": True/False}
members = {}      # member_id -> {"name":..., "age":..., "contact":..., "borrowed": [book_id, ...]}
borrow_log = []   # list of {"member_id":..., "book_id":..., "action": "BORROW"/"RETURN"}

# ---------- Helper Functions ----------

def add_book():
    print("\n--- Add New Book ---")
    book_id = input("Enter Book ID: ").strip()
    if book_id in books:
        print("Book ID already exists. Try a different ID.")
        return
    title = input("Enter Title: ").strip()
    author = input("Enter Author: ").strip()
    genre = input("Enter Genre: ").strip()

    books[book_id] = {
        "title": title,
        "author": author,
        "genre": genre,
        "available": True
    }
    print("Book added successfully!")

def add_member():
    print("\n--- Add New Member ---")
    member_id = input("Enter Member ID: ").strip()
    if member_id in members:
        print("Member ID already exists. Try a different ID.")
        return
    name = input("Enter Name: ").strip()
    age = input("Enter Age: ").strip()
    contact = input("Enter Contact Info: ").strip()

    members[member_id] = {
        "name": name,
        "age": age,
        "contact": contact,
        "borrowed": []
    }
    print("Member added successfully!")

def issue_book():
    print("\n--- Issue Book ---")
    member_id = input("Enter Member ID: ").strip()
    if member_id not in members:
        print("Member not found.")
        return

    book_id = input("Enter Book ID: ").strip()
    if book_id not in books:
        print("Book not found.")
        return

    # Check availability
    if not books[book_id]["available"]:
        print("Book is already issued to someone else.")
        return

    # Issue the book
    books[book_id]["available"] = False
    members[member_id]["borrowed"].append(book_id)
    borrow_log.append({"member_id": member_id, "book_id": book_id, "action": "BORROW"})
    print(f"Book '{books[book_id]['title']}' issued to {members[member_id]['name']}.")

def return_book():
    print("\n--- Return Book ---")
    member_id = input("Enter Member ID: ").strip()
    if member_id not in members:
        print("Member not found.")
        return

    book_id = input("Enter Book ID: ").strip()
    if book_id not in books:
        print("Book not found.")
        return

    # Check if this member actually has this book
    if book_id not in members[member_id]["borrowed"]:
        print("This member has not borrowed this book.")
        return

    # Return the book
    members[member_id]["borrowed"].remove(book_id)
    books[book_id]["available"] = True
    borrow_log.append({"member_id": member_id, "book_id": book_id, "action": "RETURN"})
    print(f"Book '{books[book_id]['title']}' returned by {members[member_id]['name']}.")

def show_available_books_by_genre():
    print("\n--- Available Books By Genre ---")
    genre = input("Enter Genre: ").strip()
    found = False
    for book_id, info in books.items():
        if info["genre"].lower() == genre.lower() and info["available"]:
            found = True
            print(f"ID: {book_id} | Title: {info['title']} | Author: {info['author']} | Genre: {info['genre']}")
    if not found:
        print("No available books found for this genre.")

def list_members_with_borrowed_books():
    print("\n--- Members Who Have Borrowed Books ---")
    found = False
    for member_id, info in members.items():
        if info["borrowed"]:
            found = True
            print(f"Member ID: {member_id} | Name: {info['name']} | Borrowed Books: {info['borrowed']}")
    if not found:
        print("No members have borrowed books currently.")

def search_book():
    print("\n--- Search Book By Title or Author ---")
    keyword = input("Enter title or author keyword: ").strip().lower()
    found = False
    for book_id, info in books.items():
        if keyword in info["title"].lower() or keyword in info["author"].lower():
            found = True
            status = "Available" if info["available"] else "Issued"
            print(f"ID: {book_id} | Title: {info['title']} | Author: {info['author']} | Genre: {info['genre']} | Status: {status}")
    if not found:
        print("No books matched your search.")

def most_popular_genre():
    print("\n--- Most Popular Genre (Based on Issued Books) ---")
    genre_count = {}
    # Count only BORROW actions
    for entry in borrow_log:
        if entry["action"] == "BORROW":
            book_id = entry["book_id"]
            genre = books[book_id]["genre"]
            genre_count[genre] = genre_count.get(genre, 0) + 1

    if not genre_count:
        print("No books have been issued yet.")
        return

    # Find genre with max count
    popular_genre = max(genre_count, key=genre_count.get)
    print(f"The most popular genre is '{popular_genre}' with {genre_count[popular_genre]} issues.")

def show_all_books():
    print("\n--- All Books ---")
    if not books:
        print("No books in library.")
        return
    for book_id, info in books.items():
        status = "Available" if info["available"] else "Issued"
        print(f"ID: {book_id} | Title: {info['title']} | Author: {info['author']} | Genre: {info['genre']} | Status: {status}")

def show_all_members():
    print("\n--- All Members ---")
    if not members:
        print("No members found.")
        return
    for member_id, info in members.items():
        print(f"ID: {member_id} | Name: {info['name']} | Age: {info['age']} | Contact: {info['contact']} | Borrowed: {info['borrowed']}")

# ---------- Main Menu Loop ----------

def main():
    while True:
        print("\n====== City Library Management System ======")
        print("1. Add Book")
        print("2. Add Member")
        print("3. Issue Book")
        print("4. Return Book")
        print("5. Show Available Books by Genre")
        print("6. List Members Who Have Borrowed Books")
        print("7. Search Book by Title/Author")
        print("8. Show Most Popular Genre")
        print("9. Show All Books")
        print("10. Show All Members")
        print("0. Exit")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            add_book()
        elif choice == "2":
            add_member()
        elif choice == "3":
            issue_book()
        elif choice == "4":
            return_book()
        elif choice == "5":
            show_available_books_by_genre()
        elif choice == "6":
            list_members_with_borrowed_books()
        elif choice == "7":
            search_book()
        elif choice == "8":
            most_popular_genre()
        elif choice == "9":
            show_all_books()
        elif choice == "10":
            show_all_members()
        elif choice == "0":
            print("Exiting... Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

# Run the program
if __name__ == "__main__":
    main()
```
