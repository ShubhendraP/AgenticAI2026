**The big idea:**  
LLMs are bad at maths and have a knowledge cutoff.  
We fix this by giving them **Tools** — Python functions the LLM can *call* when needed.

## Step 1 — Define Tools with `@tool`

`@tool` turns any Python function into something the LLM can call.  
The **docstring** is critical — the LLM reads it to decide *when* to use the tool.

```python
import math

from langchain.tools import tool
```

```python
## Tool 1 - calculator
## The docstring  will tell LLM to use it for maths

@tool
def calculator(expression: str) -> str:
  ''' Eveluate the maths expression . Input must be a Python Maths expression .
  Examples : 'math.sqrt(256)' , '2**10' , '(345*23)+99'
  '''

  result =  str(eval(expression))
  return result


# Tool Search - The LLM picks this it needs up to date factual information

@tool
def search(query:str) -> str:
  ''' Search the web for the current facts , people or events .
  Use when the question needs real world answers or recent information
  '''

  db = {
        "CEO of Google":    "Sundar Pichai is the CEO of Google.",
        "CEO of Microsoft": "Satya Nadella is the CEO of Microsoft.",
        "CEO of OpenAI":    "Sam Altman is the CEO of OpenAI.",
        "US President":     "The US President is Donald Trump (2025).",
    }

  for key in db :
    if key.lower() in query.lower() :

        return db[key]
  return f'No result found for :{query}'


  # Quick sanity check — test tools directly before plugging into an agent
print("Calculator tool test:")
print(f"  (345*23)+99 = {calculator.invoke('(345*23)+99')}")
print(f"  math.sqrt(256) = {calculator.invoke('math.sqrt(256)')}")
print(f"  math.sqrt(144) = {calculator.invoke('math.sqrt(144)')}")


print("\n🔍 Search tool test:")
print(f"  {search.invoke('CEO of Google')}")
print(f"  {search.invoke('US President')}")
```
