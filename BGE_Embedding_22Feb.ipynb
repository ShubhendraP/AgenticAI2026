```python
from transformers import AutoTokenizer, AutoModel

import torch
```

```python
#---------------
# Loading Model
#---------------
# Load model directly
#from transformers import AutoTokenizer, AutoModel

tokenizer = AutoTokenizer.from_pretrained("BAAI/bge-small-en-v1.5")
model = AutoModel.from_pretrained("BAAI/bge-small-en-v1.5")
```

```python
#--------------
# Example Sentences
#--------------

#sentences = [
 #   'I applied for a home loan at the bank',
  #  'Children were playing near the river bank'
#]


sentences = [
    'I applied for a home loan at the bank',
    'Children were playing near the home loan bank'
]
```

```python

inputs= tokenizer(
    sentences,
    padding = True,
    return_tensors = 'pt' #pt = pytorch , tensors = multi dimensional matrix
)
```

I, applied, for, a, home, loan, at, the, bank

Children, were, playing, near, the, river, bank, PAD, PAD

```python
#------------------------
#Forrward Pass
#------------------------

with torch.no_grad() :
  outputs = model(**inputs)
```

```python
embeddings = outputs.last_hidden_state[:,0,:]
```

```python
#embeddings
```

```python
#--------------
# Normalise
#--------------

embeddings = torch.nn.functional.normalize(
    embeddings
)
```

```python
#-------------
# Cosine Similarity
#-------------

from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(
    embeddings[0].unsqueeze(0),
    embeddings
)
print ("\nCosine Similarity Matrix:")
print(similarity)



```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model = SentenceTransformer('BAAI/bge-small-en-v1.5')
```

```python

#sentences = [
 #   'I applied for a home loan at the bank',
  #  'Children were playing near the river bank'
#]


sentences = [
    'I applied for a home loan at the bank',
    'Children were playing near the home loan bank'
]
```

```python
embeddings = model.encode(sentences)
```
