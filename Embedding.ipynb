# Gensim
This is a NLP based library used for word embeddings and semantix similarities.
It is trained on large datasets of wiki and news articles

```python
#pip install gensim
```

```python
import gensim.downloader as api
model = api.load('glove-wiki-gigaword-50')
```

```python
word = 'king'
model[word]
```

```python
word = 'chatgpt'
model[word]
```

```python
model ['banana']
```

```python
model.most_similar('king', topn = 5)
```

```python
model.most_similar('banana', topn = 5)
```

```python
model.most_similar('stupid', topn = 5)
```

```python
model.most_similar(
    positive = ['king', 'woman'],
    negative = ['man'],
    topn = 5
)
```

```python
# Cosine similarity

from sklearn.metrics.pairwise import cosine_similarity
cosine_similarity([model['king'],model['queen']])

```

```python
cosine_similarity([[1,0],[0,1]])
```

```python
cosine_similarity([model['king'],model['banana']])
```
